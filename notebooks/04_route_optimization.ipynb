## The Traveling Salesman Problem (TSP)

Visit all cities and return to start. Find the shortest route.

```python
from supplycm.routing import tsp_nearest_neighbor, tsp_two_opt, tsp_held_karp

# Distance matrix between 5 cities
distances = [
    [0,  10, 15, 20, 25],
    [10, 0,  35, 25, 30],
    [15, 35, 0,  30, 20],
    [20, 25, 30, 0,  15],
    [25, 30, 20, 15, 0],
]

city_names = ['Warehouse', 'Customer A', 'Customer B', 'Customer C', 'Customer D']

# Method 1: Nearest Neighbor (fast but not optimal)
route_nn, dist_nn = tsp_nearest_neighbor(distances)
print('Nearest Neighbor:')
print(f'  Route: {' -> '.join(city_names[i] for i in route_nn)}')
print(f'  Distance: {dist_nn}')
```

```python
# Method 2: 2-opt improvement (better quality)
route_2opt, dist_2opt = tsp_two_opt(distances)
print('Two-opt improved:')
print(f'  Route: {' -> '.join(city_names[i] for i in route_2opt)}')
print(f'  Distance: {dist_2opt}')
print(f'  Improvement: {((dist_nn - dist_2opt) / dist_nn * 100):.1f}%')
```

## Vehicle Routing Problem (VRP)

Multiple vehicles with capacity constraints.

```python
from supplycm.routing import vrp_capacitated_greedy, vrp_savings

# 6 customers (index 0 = depot)
demands = [0, 5, 10, 8, 12, 6]
vrp_distances = [
    [0,  5, 10, 8,  12, 6],
    [5,  0, 6,  7,  9,  4],
    [10, 6, 0,  5,  8,  7],
    [8,  7, 5,  0,  4,  6],
    [12, 9, 8,  4,  0,  5],
    [6,  4, 7,  6,  5,  0],
]

vehicle_capacity = 15
routes = vrp_capacitated_greedy(vrp_distances, demands, vehicle_capacity)

print(f'Vehicle capacity: {vehicle_capacity} units')
print(f'Number of routes: {len(routes)}')
for i, route in enumerate(routes):
    route_demand = sum(demands[j] for j in route)
    print(f'  Route {i+1}: {route} (demand: {route_demand})')
```

## Assignment Problem

Assign workers to tasks at minimum cost using the Hungarian algorithm.

```python
from supplycm.routing import assignment_problem_hungarian

# Cost matrix: worker x task
costs = [
    [10, 15, 20],
    [5,  12, 8],
    [14, 7,  11],
]

assignments, total_cost = assignment_problem_hungarian(costs)
print('Optimal assignments:')
for worker, task in assignments:
    print(f'  Worker {worker} -> Task {task} (cost: {costs[worker][task]})')
print(f'Total cost: {total_cost}')
```

## Key Takeaways

1. TSP: visit all locations, return to start
2. VRP: multiple vehicles with capacity limits
3. Nearest Neighbor is fast but suboptimal
4. 2-opt improves any route
5. Hungarian algorithm solves assignment problems optimally
