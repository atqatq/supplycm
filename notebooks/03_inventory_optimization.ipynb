# Inventory Optimization Tutorial

Learn how to determine optimal order quantities, safety stock, and reorder points.

## Scenario

You manage a warehouse for a retail product. You need to decide:
1. How much to order each time (EOQ)
2. When to place orders (reorder point)
3. How much safety stock to hold

```python
from supplycm.inventory import (
    economic_order_quantity,
    safety_stock_normal,
    reorder_point,
    abc_analysis,
    inventory_turnover_ratio,
    days_of_supply
)

# Product parameters
annual_demand = 12000   # units/year
ordering_cost = 100     # dollars per order
holding_cost = 5        # dollars per unit per year
unit_cost = 50          # purchase price per unit
lead_time = 2           # weeks
demand_std = 30         # units per week (standard deviation)
z_score = 1.96          # 97.5% service level
```

## Step 1: Economic Order Quantity (EOQ)

The optimal order size that minimizes total inventory cost.

```python
eoq = economic_order_quantity(annual_demand, ordering_cost, holding_cost)
print(f'EOQ: {eoq:.0f} units')
print(f'Orders per year: {annual_demand / eoq:.1f}')
print(f'Cycle time: {365 / (annual_demand / eoq):.0f} days')
print(f'Average inventory: {eoq / 2:.0f} units')
print(f'Annual ordering cost: ${annual_demand / eoq * ordering_cost:.2f}')
print(f'Annual holding cost: ${eoq / 2 * holding_cost:.2f}')
print(f'Total inventory cost: ${annual_demand / eoq * ordering_cost + eoq / 2 * holding_cost:.2f}')
```

## Step 2: Safety Stock

Buffer inventory to protect against demand variability during lead time.

```python
safety_stock = safety_stock_normal(z_score, demand_std, lead_time)
print(f'Z-score: {z_score} (97.5% service level)')
print(f'Demand std dev: {demand_std} units/week')
print(f'Lead time: {lead_time} weeks')
print(f'Safety stock: {safety_stock:.0f} units')
print(f'Safety stock value: ${safety_stock * unit_cost:.2f}')
```

## Step 3: Reorder Point

The inventory level that triggers a new order.

```python
demand_rate = annual_demand / 52  # weekly demand
rop = reorder_point(demand_rate, lead_time, safety_stock)
print(f'Weekly demand: {demand_rate:.0f} units')
print(f'Reorder point: {rop:.0f} units')
print(f'When inventory drops to {rop:.0f}, place an order for {eoq:.0f} units')
```

## Step 4: ABC Analysis

Not all products are equal. Classify by value contribution.

```python
products = [
    ('Product A', 500000),
    ('Product B', 300000),
    ('Product C', 100000),
    ('Product D', 50000),
    ('Product E', 30000),
    ('Product F', 15000),
    ('Product G', 5000),
]

abc = abc_analysis(products)
print(f'{'Product':<12} {'Value':>10} {'Class':>6} {'Cumulative':>10}')
print('-' * 42)
for item, cls, cum in abc:
    value = dict(products)[item]
    print(f'{item:<12} ${value:>9,} {cls:>6} {cum*100:>9.1f}%')
```

## Step 5: Performance Metrics

How efficiently is your inventory managed?

```python
cogs = annual_demand * unit_cost
avg_inventory = eoq / 2 + safety_stock
turnover = inventory_turnover_ratio(cogs, avg_inventory * unit_cost)
dos = days_of_supply(avg_inventory * unit_cost, cogs)

print(f'COGS: ${cogs:,.0f}')
print(f'Average inventory: {avg_inventory:.0f} units (${avg_inventory * unit_cost:,.0f})')
print(f'Inventory turnover: {turnover:.1f} times per year')
print(f'Days of supply: {dos:.0f} days')
```

## Summary

- EOQ tells you **how much** to order
- Safety stock protects against **uncertainty**
- Reorder point tells you **when** to order
- ABC analysis tells you **where to focus**
- Turnover ratio measures **efficiency**
