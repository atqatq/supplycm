## Available Categories

supplycm organizes algorithms into categories. Let's see what's available:

````python
categories = [attr for attr in dir(supplycm) if not attr.startswith('_')]
print(f'Total categories: {len(categories)}')
for cat in sorted(categories):
    mod = getattr(supplycm, cat)
    if hasattr(mod, '__file__'):
        algos = [a for a in dir(mod) if not a.startswith('_') and callable(getattr(mod, a))]
        print(f'  supplycm.{cat}: {len(algos)} algorithms')
````

## Your First Calculation

Let's calculate the Economic Order Quantity (EOQ), the most famous inventory formula:

````python
from supplycm.inventory import economic_order_quantity

annual_demand = 12000  # units per year
ordering_cost = 100    # dollars per order
holding_cost = 5       # dollars per unit per year

eoq = economic_order_quantity(annual_demand, ordering_cost, holding_cost)
print(f'Annual demand: {annual_demand} units')
print(f'Ordering cost: ${ordering_cost}')
print(f'Holding cost: ${holding_cost}/unit/year')
print(f'Optimal order quantity (EOQ): {eoq:.0f} units')
print(f'Number of orders per year: {annual_demand / eoq:.1f}')
print(f'Annual ordering cost: ${annual_demand / eoq * ordering_cost:.2f}')
print(f'Annual holding cost: ${eoq / 2 * holding_cost:.2f}')
print(f'Total annual cost: ${annual_demand / eoq * ordering_cost + eoq / 2 * holding_cost:.2f}')
````
