# Quality and Lean Tutorial

Learn Six Sigma quality metrics and Lean manufacturing principles.

## Six Sigma Quality Metrics

Six Sigma aims for 3.4 defects per million opportunities.

```python
from supplycm.quality import dpmo, sigma_level, process_capability_cp, process_capability_cpk

# A factory produces 10,000 units with 50 defects
defects = 50
units = 10000
opportunities = 10  # 10 ways a unit can be defective

dpmo_value = dpmo(defects, units, opportunities)
sigma = sigma_level(dpmo_value)

print(f'Defects: {defects}')
print(f'Units: {units}')
print(f'Opportunities per unit: {opportunities}')
print(f'DPMO: {dpmo_value:.1f}')
print(f'Sigma level: {sigma:.1f}')
print()
print('Six Sigma target: 3.4 DPMO = 6.0 sigma')
```

## Process Capability

Can your process meet specifications?

```python
# Product specification: 10 +/- 2 (USL=12, LSL=8)
USL = 12
LSL = 8
mean = 10.2  # slightly off-center
std_dev = 0.5

cp = process_capability_cp(USL, LSL, std_dev)
cpk = process_capability_cpk(USL, LSL, mean, std_dev)

print(f'Specification: {LSL} to {USL}')
print(f'Process mean: {mean}')
print(f'Std dev: {std_dev}')
print(f'Cp:  {cp:.2f} (potential capability)')
print(f'Cpk: {cpk:.2f} (actual capability)')
print()
print('Interpretation:')
if cpk >= 1.33:
    print('  Process is capable (Cpk >= 1.33)')
elif cpk >= 1.0:
    print('  Process is marginally capable (1.0 <= Cpk < 1.33)')
else:
    print('  Process is not capable (Cpk < 1.0)')
```

## Lean Manufacturing Metrics

### Takt Time: The pace of customer demand

```python
from supplycm.lean import takt_time, oee, wip_calculation

# Available: 8 hours = 480 minutes, minus 60 min breaks = 420 min
available_time = 420  # minutes
customer_demand = 210  # units per day

takt = takt_time(available_time, customer_demand)
print(f'Available time: {available_time} min/day')
print(f'Customer demand: {customer_demand} units/day')
print(f'Takt time: {takt:.1f} min/unit')
print(f'Produce one unit every {takt:.1f} minutes to meet demand')
```

### OEE: Overall Equipment Effectiveness

```python
# Machine data
availability = 0.85   # 85% of scheduled time running
performance = 0.90    # 90% of ideal speed
quality = 0.95        # 95% good parts

oee_value = oee(availability, performance, quality)
print(f'Availability: {availability*100:.0f}%')
print(f'Performance: {performance*100:.0f}%')
print(f'Quality: {quality*100:.0f}%')
print(f'OEE: {oee_value*100:.1f}%')
print()
print('World class OEE: 85%')
print('Typical OEE: 60%')
if oee_value >= 0.85:
    print('Status: World class!')
elif oee_value >= 0.60:
    print('Status: Typical')
else:
    print('Status: Room for improvement')
```

### Little's Law: WIP = Throughput x Flow Time

```python
throughput = 10  # units per hour
flow_time = 5    # hours per unit

wip = wip_calculation(throughput, flow_time)
print(f'Throughput: {throughput} units/hour')
print(f'Flow time: {flow_time} hours')
print(f'WIP (Work In Progress): {wip} units')
print()
print('Little\'s Law: WIP = Throughput x Flow Time')
print('If you produce 10/hr and each takes 5hr, you have 50 in progress.')
```

## Key Takeaways

1. DPMO and sigma level measure quality
2. Cp/Cpk measure process capability
3. Takt time sets the production pace
4. OEE measures equipment effectiveness
5. Little's Law relates WIP, throughput, and flow time
