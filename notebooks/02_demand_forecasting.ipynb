# Demand Forecasting Tutorial

Learn how to forecast demand using different methods in supplycm.

## The Problem

You have 12 months of historical demand data. You need to forecast the next 3 months. Which method should you use?

```python
from supplycm.forecasting import (
    simple_moving_average,
    single_exponential_smoothing,
    holt_winters,
    naive_forecast
)
from supplycm.statistics import mape

# 12 months of demand data
demand = [100, 120, 115, 130, 125, 140, 150, 145, 160, 155, 170, 175]
print(f'Historical demand (12 months): {demand}')
```

## Method 1: Naive Forecast

The simplest method: next month equals this month.

```python
naive = naive_forecast(demand)
print(f'Naive forecast for month 13: {naive[-1]}')
```

## Method 2: Simple Moving Average (SMA)

Average the last N months. Smooths out noise.

```python
sma_3 = simple_moving_average(demand, window=3)
sma_6 = simple_moving_average(demand, window=6)
print(f'SMA-3 forecast: {sma_3[-1]:.1f}')
print(f'SMA-6 forecast: {sma_6[-1]:.1f}')
```

## Method 3: Single Exponential Smoothing (SES)

Weight recent data more heavily. Alpha controls responsiveness.

```python
ses_low = single_exponential_smoothing(demand, alpha=0.1)  # smooth
ses_high = single_exponential_smoothing(demand, alpha=0.5)  # responsive
print(f'SES (alpha=0.1): {ses_low[-1]:.1f}')
print(f'SES (alpha=0.5): {ses_high[-1]:.1f}')
```

## Method 4: Holt-Winters (Triple Exponential Smoothing)

Captures trend and seasonality. Best for data with patterns.

```python
level, trend, seasonal = holt_winters(
    demand, alpha=0.5, beta=0.1, gamma=0.1, season_length=4
)
print(f'Final level: {level[-1]:.1f}')
print(f'Final trend: {trend[-1]:.1f}')
print(f'Seasonal pattern: {[round(s, 1) for s in seasonal[-4:]]}')
```

## Comparing Methods

Which method is most accurate? Use MAPE (Mean Absolute Percentage Error).

```python
# Use last 3 months as test data
train = demand[:-3]
test = demand[-3:]

# Forecast using each method
from supplycm.forecasting import average_method
f_naive = [train[-1]] * 3
f_sma = [sum(train[-3:]) / 3] * 3
f_ses = [single_exponential_smoothing(train, alpha=0.3)[-1]] * 3

print(f'Actual:      {test}')
print(f'Naive:       {f_naive}')
print(f'SMA-3:       {[round(f, 1) for f in f_sma]}')
print(f'SES:         {[round(f, 1) for f in f_ses]}')
print()
print(f'MAPE Naive:  {mape(test, f_naive):.1f}%')
print(f'MAPE SMA:    {mape(test, f_sma):.1f}%')
print(f'MAPE SES:    {mape(test, f_ses):.1f}%')
```

## Key Takeaways

1. Start with naive forecast as a baseline
2. Use SMA for stable demand
3. Use SES for trending demand
4. Use Holt-Winters for seasonal demand
5. Always measure accuracy with MAPE or similar metric
